```python
from pathlib import Path

# Folder to search (Windows path)
folder = Path(r"C:\Users\manue\Music\fg\FG")
```

```python

# Text to append
text_to_add = " * FG"

# Loop through all .txt files in that folder
for file in folder.glob("*.mp3"):
    new_name = file.stem + text_to_add + file.suffix
    new_path = file.with_name(new_name)
    file.rename(new_path)
    print(f"Renomeado: {file.name} → {new_name}")
```
